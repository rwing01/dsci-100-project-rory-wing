I plan to use a KNN classification to predict a team's tournament finish based on its regional win percentage and tournament seed. I believe this method is good because tournament finish is a categorical variable, and KNN classification predicts catergories identifiying similar observations based on their predictor values. However, KNN can be sensitive to the number of neighbours selected and differences in predictor scales. Additionally, some tournament finishes occur much more frequently than others, which could make it harder to accurately predict less common outcomes.

Before building the model, I will ensure observations with missing values are removed in relevant columns and split the data into a 3:1 training and testing split, using strata() based on tournament finish to help maintain class proportions. I will then use recipe() to to center and scale both predictors so they contribute equally(ish) to distance calculations. Next, I will perform a 5-fold cross validiation on the training data to select the optimal K, without using a separate validation set.

I will assess model performance using classification accuracy, which measures the proportion of correctly predicted tournament finishes. I will also examine a confusion matrix to identify which tournament outcomes are most frequently misclassified. Finally, I will evaluate the selected model on the testing set to estimate how accurately it predicts unseen data.

#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
library(tidyverse)
library(tidymodels)
set.seed(1234)

#Keep predictors + response and remove missing values
march_model_data <- march_madness_tidy |>
    select(reg_wins_pct, seed, tourney_finish) |>
    drop_na() |>
    mutate(tourney_finish = as_factor(tourney_finish))

#Check the number of observations in each category
march_model_data |>
    count(tourney_finish)

#Split into 3:1 training/testing
march_split <- initial_split(march_model_data, prop = 0.75, strata = tourney_finish)
march_training <- training(march_split)
march_testing <- testing(march_split)

#Specifty respomse and two predictors
march_recipe <- recipe(tourney_finish ~ reg_wins_pct + seed, data = march_training) |>
    step_scale(all_predictors()) |>
    step_center(all_predictors()) 

#Create KNN Classification model
knn_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |>
    set_engine("kknn") |>
    set_mode("classification")

#Create five cross-validation folds
march_vfold <- vfold_cv(march_training, v = 5, strata = tourney_finish)

#Combine the model and recipe
march_workflow <- workflow() |>
    add_recipe(march_recipe) |>
    add_model(knn_spec)

#Try different values of K
k_values <- tibble(neighbors = seq(from = 1, to = 51, by = 5))

#Calculate cross_validation performance
march_results <- march_workflow |>
    tune_grid(resamples = march_vfold, grid = k_values) |>
    collect_metrics()

#Keep accuracy results -> K = 41 turned out to be the best
march_accuracies <- march_results |>
    filter(.metric == "accuracy")
march_accuracies

#Selecting best K
best_k <- march_accuracies |>
    arrange(desc(mean)) |>
    head(1) |>
    pull(neighbors)
best_k

#Final KNN Model
march_best_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = best_k) |>
    set_engine("kknn") |>
    set_mode("classification")

#Fit model using training data
march_fit <- workflow() |>
    add_recipe(march_recipe) |>
    add_model(march_best_spec) |>
    fit(data = march_training)

#Predict tourney_finish
march_predictions <- march_fit |>
    predict(new_data = march_testing) |>
    bind_cols(march_testing)

#Calculate test accuracy
march_metrics <- march_predictions |>
    metrics(truth = tourney_finish, estimate = .pred_class)

march_metrics

#Create confusion matrix
march_conf_mat <- march_predictions |>
    conf_mat(truth = tourney_finish, estimate = .pred_class)
march_conf_mat
    
```

Cross-validation is important because it helps us select a suitable K using the training data without using the testing set. Accuracy generally is increased as K increased although some fluctuations occured. After the final model was fitted, I achieved a test accuracy of about 55.15%, suggesting that the model has sokme predictive ability but still makes a considerable number of incorrect classifications.

```r
#cross-validation accuracy vs different K values plot
options(repr.plot.height = 6, repr.plot.width = 8)
options(jupyter.plot_mimetypes = c("text/plain", "image/png"))
march_accuracy_plot <- march_accuracies |>
    ggplot(aes(x = neighbors, y = mean)) +
    geom_point() +
    geom_line() +
    labs(title = "KNN Cross-Validation Accuracy", x = "Number of Neighbors (K)", y = "Cross-Validation Accuracy") + 
    theme(text = element_text(size = 12))
march_accuracy_plot
```
